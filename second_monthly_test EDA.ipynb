```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the plotting grid
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Boxplots of Numerical Columns to Identify Outliers', fontsize=16)

# List of numerical columns to plot
numerical_cols = ['Quantity', 'Unit_Price', 'Discount', 'Sales']

for ax, col in zip(axes.flatten(), numerical_cols):
    sns.boxplot(x=df[col], ax=ax, palette='Set2')
    ax.set_title(f'Boxplot of {col}', fontsize=12)
    ax.set_xlabel('')

plt.tight_layout()
plt.show()
```

PART B

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot a histogram of the Sales column to visualize its distribution
plt.figure(figsize=(10, 6))
sns.histplot(df['Sales'], bins=50, kde=True, color='skyblue')
plt.title('Distribution of Sales (Histogram)', fontsize=14)
plt.xlabel('Sales', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate average sales per category
avg_sales_by_category = df.groupby('Category')['Sales'].mean().reset_index().sort_values(by='Sales', ascending=False)

# Plot the average sales in a bar chart
plt.figure(figsize=(10, 6))
sns.barplot(x='Category', y='Sales', data=avg_sales_by_category, palette='viridis')
plt.title('Average Sales by Category', fontsize=14)
plt.xlabel('Category', fontsize=12)
plt.ylabel('Average Sales ($)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group by category and sum up total sales
sales_by_category = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

# Plotting the pie chart
plt.figure(figsize=(8, 8))
colors = sns.color_palette('pastel')[0:len(sales_by_category)]

plt.pie(
    sales_by_category,
    labels=sales_by_category.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=colors,
    wedgeprops={'edgecolor': 'white', 'linewidth': 1.5}
)

plt.title('Sales Contribution by Category', fontsize=14)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group by category, calculate total sales, and sort from lowest to highest
sales_trend = df.groupby('Category')['Sales'].sum().sort_values(ascending=True).reset_index()

# Plot a line chart to show the progression from the lowest performing to the highest
plt.figure(figsize=(10, 6))
sns.lineplot(x='Category', y='Sales', data=sales_trend, marker='o', linewidth=2.5, color='royalblue')

plt.title('Ascending Performance of Categories (Lowest to Highest Sales)', fontsize=14)
plt.xlabel('Category', fontsize=12)
plt.ylabel('Total Sales ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a scatter plot representing the main numerical variables:
# Unit_Price on x-axis, Sales on y-axis, sized by Quantity, and colored by Discount
plt.figure(figsize=(12, 7))
sns.scatterplot(
    data=df,
    x='Unit_Price',
    y='Sales',
    hue='Discount',
    size='Quantity',
    sizes=(20, 200),
    palette='viridis',
    alpha=0.7
)

plt.title('Distribution and Relationship of Important Numerical Values', fontsize=14)
plt.xlabel('Unit Price ($)', fontsize=12)
plt.ylabel('Sales ($)', fontsize=12)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title='Legend')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a scatter plot representing the relationship between key numerical variables
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x='Unit_Price',
    y='Sales',
    hue='Discount',
    palette='viridis',
    alpha=0.7
)

plt.title('Relationship Between Unit Price and Sales', fontsize=14)
plt.xlabel('Unit Price ($)', fontsize=12)
plt.ylabel('Sales ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(title='Discount')
plt.show()
```
